## 1. NumPy Basics

```python
import numpy as np

# Create arrays
a = np.array([1, 2, 3, 4, 5])
b = np.zeros((3, 3))
c = np.ones((2, 4))
d = np.arange(0, 10, 2)   # [0 2 4 6 8]
e = np.linspace(0, 1, 5)  # 5 evenly-spaced values between 0 and 1

print('a:', a)
print('zeros:\n', b)
print('ones:\n', c)
print('arange:', d)
print('linspace:', e)
```

```python
# Array operations
x = np.array([10, 20, 30, 40])
y = np.array([1, 2, 3, 4])

print('Addition:', x + y)
print('Element-wise multiply:', x * y)
print('Dot product:', np.dot(x, y))
print('Mean:', x.mean(), '  Std:', x.std())
print('Max:', x.max(), '  Min:', x.min())
```

```python
# Reshaping & slicing
m = np.arange(1, 13).reshape(3, 4)
print('Matrix:\n', m)
print('Row 0:', m[0])
print('Column 1:', m[:, 1])
print('Sub-matrix (rows 0-1, cols 1-2):\n', m[:2, 1:3])
```

```python
# Random numbers (important for ML reproducibility)
np.random.seed(42)
rand_arr = np.random.randn(3, 3)   # standard normal
rand_int = np.random.randint(0, 10, size=(2, 5))
print('Normal random:\n', rand_arr)
print('Int random:\n', rand_int)
```

## 2. Pandas Basics

```python
import pandas as pd

# Create a DataFrame
data = {
    'Name':   ['Alice', 'Bob', 'Carol', 'Dave', 'Eve'],
    'Age':    [24, 27, 22, 32, 29],
    'Score':  [88.5, 92.0, 75.3, 95.1, 81.7],
    'Passed': [True, True, False, True, True]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Basic info
print(df.shape)    # rows, columns
print(df.dtypes)   # column types
df.describe()      # summary statistics
```

```python
# Selecting & filtering
print(df['Age'])                       # single column
print(df[['Name', 'Score']])           # multiple columns
print(df[df['Score'] > 85])            # filter rows
print(df.loc[0:2, 'Name':'Score'])     # label-based slice
```

```python
# Handling missing values
df_missing = df.copy()
df_missing.loc[1, 'Score'] = None
df_missing.loc[3, 'Age']   = None

print('Missing values per column:\n', df_missing.isnull().sum())

df_filled  = df_missing.fillna({'Score': df_missing['Score'].mean(), 'Age': 0})
df_dropped = df_missing.dropna()
print('After fill:\n', df_filled)
print('After drop:\n', df_dropped)
```

```python
# GroupBy & aggregation
df['Grade'] = pd.cut(df['Score'], bins=[0, 80, 90, 100], labels=['C', 'B', 'A'])
print(df.groupby('Grade')['Score'].agg(['mean', 'count']))
```
