# 02 – Data Visualization for Machine Learning

Good visualizations reveal patterns before you train any model.

Libraries covered:
- **Matplotlib** – low-level plotting
- **Seaborn** – statistical plots built on Matplotlib

---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style='whitegrid')
np.random.seed(42)
```

## 1. Matplotlib – Core Plots

```python
# Line plot
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y1, label='sin(x)', color='steelblue')
plt.plot(x, y2, label='cos(x)', color='coral', linestyle='--')
plt.title('Line Plot – sin & cos')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Scatter plot
x_scatter = np.random.randn(100)
y_scatter = 2 * x_scatter + np.random.randn(100) * 0.5

plt.figure(figsize=(6, 5))
plt.scatter(x_scatter, y_scatter, alpha=0.6, edgecolors='k', linewidths=0.5)
plt.title('Scatter Plot')
plt.xlabel('Feature X')
plt.ylabel('Target Y')
plt.show()
```

```python
# Bar & histogram side-by-side
categories = ['Math', 'Science', 'English', 'History']
scores     = [85, 92, 78, 88]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].bar(categories, scores, color=['steelblue', 'coral', 'green', 'purple'])
axes[0].set_title('Bar Chart')
axes[0].set_ylabel('Score')

data = np.random.normal(70, 10, 200)
axes[1].hist(data, bins=20, color='steelblue', edgecolor='k', alpha=0.7)
axes[1].set_title('Histogram')
axes[1].set_xlabel('Value')

plt.tight_layout()
plt.show()
```

## 2. Seaborn – Statistical Plots

```python
# Use Iris dataset (built-in to seaborn)
iris = sns.load_dataset('iris')
iris.head()
```

```python
# Pair plot – relationships between all numeric features
sns.pairplot(iris, hue='species', palette='husl')
plt.suptitle('Iris Pair Plot', y=1.02)
plt.show()
```

```python
# Box plot & Violin plot
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(data=iris, x='species', y='petal_length', palette='pastel', ax=axes[0])
axes[0].set_title('Box Plot – Petal Length by Species')

sns.violinplot(data=iris, x='species', y='petal_length', palette='muted', ax=axes[1])
axes[1].set_title('Violin Plot – Petal Length by Species')

plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap – critical for feature selection
corr = iris.drop(columns='species').corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', square=True, linewidths=0.5)
plt.title('Feature Correlation Heatmap')
plt.show()
```

```python
# Count plot – useful for class imbalance check
sns.countplot(data=iris, x='species', palette='Set2')
plt.title('Class Distribution')
plt.show()
```

## 3. Key Takeaways

| Plot type | Use case |
|-----------|----------|
| Line | Trends over time / epochs |
| Scatter | Relationship between two features |
| Histogram | Distribution of a single feature |
| Box / Violin | Feature distribution per class |
| Heatmap | Feature correlations |
| Pair plot | All pairwise feature relationships |
